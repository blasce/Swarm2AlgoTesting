```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import re
import ast

with open("Coat_GA_1_sizes_2_trials_gran_10_2625757929.log", "r") as f:
    text = f.read()


match = re.search(r'string:\s*(\[\[\[.*?\]\]\])', text, re.DOTALL)
if match:
    array_text = match.group(1)
    array = ast.literal_eval(array_text)
    print(array)

data = np.array(array)

fig, axes = plt.subplots(1,4, figsize=(16, 4))

for j, ax in enumerate(axes):
    sns.heatmap(data[j], cmap="viridis", ax = ax, cbar=(j==3))
```

```python
matches = re.findall(r'string:\s*(\[\[\[.*?\]\]\])', text, re.DOTALL)

arrays = [ast.literal_eval(m) for m in matches]
print(f"Found {len(arrays)} arrays")
```

```python
interval_selected = []

for i, arr in enumerate(arrays):
    if i % 15 == 0:
        interval_selected.append(arr)
for i in interval_selected:
    data = np.array(i)

    fig, axes = plt.subplots(1, 4, figsize=(16,4))
    for j, ax in enumerate(axes):
        sns.heatmap(data[j], cmap="viridis", ax=ax, cbar=(j==3))
    
    plt.tight_layout()
    plt.show()
```
