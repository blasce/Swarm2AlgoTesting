```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns



genome = [[[0.15553, 0.70551, 0.49352, 0.78322], [0.10006, 0.85780, 0.05751, 0.32809], [0.17741, 0.31106, 0.92943, 0.24894]], [[0.23129, 0.87715, 0.99753, 0.69382], [0.05476, 0.74466, 0.83709, 0.78959], [0.26128, 0.98465, 0.12287, 0.90565]], [[0.60923, 0.29582, 0.95400, 0.85113], [0.00361, 0.45711, 0.35224, 0.86309], [0.85265, 0.79387, 0.46479, 0.17105]], [[0.64470, 0.43506, 0.20252, 0.44396], [0.30038, 0.08123, 0.87981, 0.06085], [0.96094, 0.00553, 0.33884, 0.67117]]]

data = np.array(genome)

fig, axes = plt.subplots(1,4, figsize=(16, 4))

for i, ax in enumerate(axes):
    sns.heatmap(data[i], cmap="viridis", ax = ax, cbar=(i==3))

plt.tight_layout()
plt.show()
```
